```python
students = [
    {"name": "Ilay", "age": 13, "grade": "A"},
    {"name": "Maya", "age": 12, "grade": "B"},
    {"name": "David", "age": 14, "grade": "A"}
]

print("Student names:")
for student in students:
    print(student["name"])

total_age = 0
for student in students:
    total_age += student["age"]
average_age = total_age / len(students)
print(f"Average age: {average_age}")

count_A = 0
for student in students:
    if student["grade"] == "A":
        count_A += 1
print(f"Number of students with grade 'A': {count_A}")
```

```python
inventory = {
    "apple": {"quantity": 50, "price": 0.5},
    "banana": {"quantity": 30, "price": 0.3},
    "orange": {"quantity": 20, "price": 0.7}
}

print("Current Inventory:")
for item, info in inventory.items():
    print(f"{item} - Quantity: {info['quantity']}, Price: ${info['price']}")

item_to_buy = input("Which item would you like to buy? ").lower()
quantity_to_buy = int(input("How many would you like to buy? "))

if item_to_buy in inventory:
    if quantity_to_buy <= inventory[item_to_buy]["quantity"]:
        inventory[item_to_buy]["quantity"] -= quantity_to_buy

        total_cost = quantity_to_buy * inventory[item_to_buy]["price"]
        print(f"Total cost: ${total_cost}")
        print(f"Updated quantity of {item_to_buy}: {inventory[item_to_buy]['quantity']}")
    else:
        print(f"Not enough {item_to_buy} in stock!")
else:
    print("Item not found in inventory!")
```

```python
gradebook = {}

def add_student(name):
    if name not in gradebook:
        gradebook[name] = []
        print(f"Student {name} added.")
    else:
        print(f"Student {name} already exists.")

def add_score(name, score):
    if name in gradebook:
        gradebook[name].append(score)
        print(f"Added score {score} for {name}.")
    else:
        print(f"Student {name} not found.")

def average_score(name):
    if name in gradebook and gradebook[name]:
        return sum(gradebook[name]) / len(gradebook[name])
    else:
        return 0

def top_student():
    if not gradebook:
        return None
    best = max(gradebook, key=lambda student: average_score(student))
    return best, average_score(best)

add_student("Ilay")
add_student("Maya")
add_student("David")

add_score("Ilay", 90)
add_score("Ilay", 85)
add_score("Maya", 95)
add_score("David", 80)
add_score("David", 85)

for student in gradebook:
    print(f"{student}'s average: {average_score(student)}")

best_student, best_avg = top_student()
print(f"Top student: {best_student} with average {best_avg}")
```
