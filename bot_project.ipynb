```python
import random



bot_name = "SmartBot"

greetings = ["Hi!", "Hello!", "Hey!", "Nice to see you!", "Hey there!"]
goodbyes = ["Bye!", "See you!", "Goodbye!", "Have a great day!", "See you soon!"]
jokes = [
    "Why did the computer go to the doctor? Because it had a virus!",
    "Why do programmers prefer dark mode? Because light attracts bugs!",
    "Why did the math book look sad? Because it had too many problems.",
    "Why was the computer cold? It forgot to close Windows!",
    "Why did the student eat his homework? Because the teacher said it was a piece of cake!"
]


def print_separator():
    print("==========================================")

def print_bot(message):
    print(f"🤖 {bot_name}: {message}")

def show_help():
    print_bot("I can do the following:")
    print("- Chat with you")
    print("- Tell jokes")
    print("- Play a guessing game")
    print("- Recognize your mood")
    print("- Do simple math")


def greet_user():
    print_bot("Hello! Welcome to my chatbot.")
    user_name = input("What's your name? ")
    print_bot(f"Nice to meet you, {user_name}!")
    return user_name

# 
# 

def tell_joke():
    return random.choice(jokes)

def play_guess_game():
    number_to_guess = random.randint(1, 20)
    attempts = 0
    print_bot("I am thinking of a number between 1 and 20. Can you guess it?")
    
    while True:
        try:
            guess = int(input("Your guess: "))
            attempts += 1
            if guess < number_to_guess:
                print_bot("Too low! Try again.")
            elif guess > number_to_guess:
                print_bot("Too high! Try again.")
            else:
                return f"🎉 Correct! You guessed it in {attempts} attempts."
        except ValueError:
            print_bot("Please enter a valid number.")

def analyze_mood(message):
    happy_words = ["happy", "great", "awesome", "good", "fun", "yay"]
    sad_words = ["sad", "bad", "terrible", "ugh", "unhappy", "angry"]
    
    message_lower = message.lower()
    
    for word in happy_words:
        if word in message_lower:
            return "happy"
    for word in sad_words:
        if word in message_lower:
            return "sad"
    return "neutral"


def get_response(message, user_name):
    message_lower = message.lower()
    
    
    if any(g in message_lower for g in ["hello", "hi", "hey"]):
        return f"{random.choice(greetings)}, {user_name}!"
    
    elif "how are you" in message_lower:
        return "I'm great! How are you?"
    
    elif "your name" in message_lower or "who are you" in message_lower:
        return f"My name is {bot_name}."
    
    elif "joke" in message_lower or "funny" in message_lower:
        return tell_joke()
    
    elif "game" in message_lower or "play" in message_lower:
        return "game_menu"
    
    
    try:
       
        result = eval(message)
        return f"The answer is: {result}"
    except:
        pass 
    
    mood = analyze_mood(message)
    if mood == "happy":
        return "I see you're happy! 😄"
    elif mood == "sad":
        return "Oh, I'm sorry you're feeling sad. 😢"
    

    default_responses = ["Interesting!", "Tell me more!", "I see...", "Cool!", f"Got it, {user_name}!"]
    return random.choice(default_responses)



def chat():
    user_name = greet_user()
    show_help()
    message_count = 0
    
    while True:
        user_message = input(f"\n{user_name}: ").strip()
        if not user_message:
            continue
        
        message_count += 1
        
        
        if any(word in user_message.lower() for word in ["bye", "goodbye", "quit", "exit"]):
            print_bot(f"{random.choice(goodbyes)}, {user_name}!")
            print_separator()
            print_bot(f"You sent {message_count} messages. Thanks for chatting!")
            print_separator()
            break
        
        response = get_response(user_message, user_name)
        
      
        if response == "game_menu":
            print_bot("What would you like to play?")
            print("1. Number Guessing Game")
            print("2. Never mind")
            choice = input("Enter choice: ").strip()
            if choice == "1":
                result = play_guess_game()
                print_bot(result)
            else:
                print_bot("Okay, maybe next time!")
        else:
            print_bot(response)


if __name__ == "__main__":
    chat()
```
