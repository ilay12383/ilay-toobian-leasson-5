```python
def safe_divide(a, b):
    try:
        result = a / b
        return result
    except ZeroDivisionError:
        return "Error: Cannot divide by zero."



print(safe_divide(10, 2))   
print(safe_divide(8, 4))    
print(safe_divide(5, 0))    
print(safe_divide(20, 5))   
```

```python
try:
    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))

    operation = input("Enter operation (+, -, *, /): ")

    if operation == "+":
        result = num1 + num2
    elif operation == "-":
        result = num1 - num2
    elif operation == "*":
        result = num1 * num2
    elif operation == "/":
        if num2 == 0:
            print("Error: Cannot divide by zero.")
        else:
            result = num1 / num2
            print("Result:", result)
    else:
        print("Error: Invalid operation.")

    if operation in ["+", "-", "*"]:
        print("Result:", result)

except ValueError:
    print("Error: Please enter valid numbers.")
```

```python
students = {}

def add_student(name):
    try:
        if name in students:
            return "Error: Student already exists."
        students[name] = []
        return f"Student {name} added."
    except Exception:
        return "Error: Could not add student."

def add_grade(name, grade):
    try:
        if name not in students:
            return "Error: Student not found."

        grade = float(grade)

        if grade < 0 or grade > 100:
            return "Error: Grade must be between 0 and 100."

        students[name].append(grade)
        return "Grade added successfully."

    except ValueError:
        return "Error: Invalid grade input."

def get_average(name):
    try:
        if name not in students:
            return "Error: Student not found."

        grades = students[name]

        if len(grades) == 0:
            return "Error: No grades available."

        return sum(grades) / len(grades)

    except Exception:
        return "Error: Could not calculate average."

def get_student_info(name):
    try:
        if name not in students:
            return "Error: Student not found."

        return {
            "name": name,
            "grades": students[name],
            "average": get_average(name)
        }

    except Exception:
        return "Error: Could not get student info."



print(add_student("Ilay"))
print(add_student("Ilay"))  
print(add_grade("Ilay", 95))
print(add_grade("Ilay", 110))  
print(add_grade("Ilay", "abc"))  

print(get_average("Ilay"))
print(get_average("Noam")) 

print(get_student_info("Ilay"))
```

```python
def list_statistics(lst):
    try:
        if len(lst) == 0:
            return "Error: List is empty."

        clean_list = []

        for item in lst:
            try:
                clean_list.append(float(item))
            except ValueError:
                return "Error: List contains non-numeric values."

        total = sum(clean_list)
        average = total / len(clean_list)
        minimum = min(clean_list)
        maximum = max(clean_list)

        return {
            "sum": total,
            "average": average,
            "min": minimum,
            "max": maximum
        }

    except Exception:
        return "Error: Something went wrong."


# Test
print(list_statistics([10, 20, 30]))
print(list_statistics([]))
print(list_statistics([10, "abc", 30]))
```

```python
import ast

def safe_parse(text):
    try:
        result = ast.literal_eval(text)

        if not isinstance(result, dict):
            return "Error: Not a valid dictionary format."

        return result

    except (ValueError, SyntaxError):
        return "Error: Invalid format."

# Test
print(safe_parse("{'name': 'Ilay', 'age': 12}"))
print(safe_parse("{name: Ilay}"))
print(safe_parse("[1,2,3]"))
```
