```python
import math


num = float(input("Enter a number: "))

print("Square root:", math.sqrt(num))
print("Number squared:", num ** 2)
print("Rounded up (ceil):", math.ceil(num))
print("Rounded down (floor):", math.floor(num))
```

```python
import random


number = random.randint(1, 100)

guess = int(input("Guess a number between 1 and 100: "))

if guess == number:
    print("Correct! 🎉")
else:
    print("Wrong 😅")
    print("The correct number was:", number)
```

```python
import turtle

t = turtle.Turtle()

t.pencolor("blue")
t.pensize(3)

for i in range(4):
    t.forward(150)
    t.right(90)

turtle.done()
```

```python
import turtle

t = turtle.Turtle()

t.pencolor("red")
for i in range(3):
    t.forward(100)
    t.left(120)

t.penup()
t.goto(200, 0)
t.pendown()

t.pencolor("blue")
for i in range(4):
    t.forward(100)
    t.right(90)

t.penup()
t.goto(-200, 0)
t.pendown()

t.pencolor("green")
t.circle(50)

turtle.done()
```

```python
import turtle

t = turtle.Turtle()

colors = ["red", "blue", "green", "yellow", "purple", "orange", "pink", "brown"]

for i in range(8):
    t.pencolor(colors[i])  
    
    for j in range(4):
        t.forward(100)
        t.right(90)
    
    t.right(45) 

turtle.done()
```

```python
import turtle

t = turtle.Turtle()

def draw_star(size, color):
    t.fillcolor(color)
    t.begin_fill()
    
    for i in range(5):
        t.forward(size)
        t.right(144)
    
    t.end_fill()

draw_star(100, "red")

t.penup()
t.goto(-150, -50)
t.pendown()

draw_star(70, "blue")

t.penup()
t.goto(150, -80)
t.pendown()

draw_star(120, "green")

turtle.done()
```

```python
import math

radius = float(input("Enter radius: "))

circumference = 2 * math.pi * radius
area = math.pi * radius ** 2

print("Circumference:", circumference)
print("Area:", area)
```
